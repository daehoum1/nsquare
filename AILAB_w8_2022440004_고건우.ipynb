## 1. Introduction: XGBoost, MLP, and Imbalanced Classification

SECOM contains semiconductor process measurements, missing sensor values, and pass/fail labels. Failures are rare. We use the sensor matrix only; timestamps in the label file are excluded. This exercise uses a random stratified split, so it does not measure generalization to future production periods.

### 1.1 Gradient-boosted trees

XGBoost adds decision trees sequentially to reduce a regularized objective:

$$F_M(x)=\sum_{m=1}^{M} f_m(x),\qquad p(y=1\mid x)=\sigma(F_M(x)).$$

The binary logistic objective learns scores; **Macro-F1 is the model-selection target**, not the differentiable training loss. Increasing the positive-class weight changes the training loss; changing the threshold changes the final decision rule.

| Setting | Role |
|:--|:--|
| `n_estimators`, `learning_rate` | Number of trees and update size |
| `max_depth`, `min_child_weight` | Tree complexity |
| `subsample`, `colsample_bytree` | Row/feature sampling |
| `reg_alpha`, `reg_lambda` | Regularization |
| `scale_pos_weight` | Weight of positive (fail) examples |
| Decision threshold | Trade-off between missed failures and false alarms |

Trees do not require standardization here. XGBoost can handle `NaN` directly; the provided helpers also support mean/median imputation from `제조데이터.ipynb`. These are available options, not required comparisons.

### 1.2 Multilayer Perceptron (MLP)
An MLP maps standardized sensor vectors through fully connected hidden layers with nonlinear activations to a single fail logit. The supplied PyTorch baseline uses `BCEWithLogitsLoss`; `pos_weight` can increase the contribution of failures. Sigmoid converts logits to scores for thresholding. Dropout and weight decay help control overfitting.

For MLP, impute missing values and fit a `StandardScaler` on Train only. Use the same fitted transformations on Validation and Test. Unlike the tree baseline, this MLP requires finite inputs. You may change the architecture, loss (for example weighted BCE or focal loss), optimizer, batch size, learning rate, epoch count, checkpoint selection, and other training choices. Selection always uses Validation; updates always use Train.

### 1.3 Why Accuracy alone is insufficient

For fail as the positive class:

$$\mathrm{Accuracy}=\frac{TP+TN}{TP+TN+FP+FN},\quad
P=\frac{TP}{TP+FP},\quad R=\frac{TP}{TP+FN},\quad F1=\frac{2PR}{P+R}.$$

$$\mathrm{Macro\!\! -\!\! F1}=\frac{F1_{pass}+F1_{fail}}{2},\qquad
\mathrm{Balanced\ Accuracy}=\frac{Recall_{pass}+Recall_{fail}}{2}.$$

- **Accuracy** gives every sample equal weight; the majority class dominates the total.
- **Macro-F1** gives the two class-specific F1 values equal weight. **Maximize this score on Validation.** It still depends on errors and class prevalence; always inspect fail recall too.
- **Fail Recall** measures how many true failures are detected. **Fail Precision** measures how often a failure alarm is correct.
- **Average Precision (AP)** summarizes the precision–recall ranking from continuous scores. It is supplementary here; it is not Macro-F1 and is not the trapezoidal PR-AUC.

**생각해 보기:** 모두 정상으로 예측하면 Accuracy는 높을 수 있지만, 불량 Recall과 F1은 0입니다. Macro-F1도 0이 되는지 직접 확인하세요. False Negative(불량 누락)와 False Positive(정상 오탐)의 현장 비용은 어떻게 다를까요?

## 3. Baselines and Reusable Evaluation

The supplied helpers retain feature names, drop all-missing/constant columns using Train only, and optionally fit a mean/median imputer on Train. Native mode retains `NaN` for XGBoost. Keep metric definitions unchanged.

Both baselines use the same Train/Validation rows. MLP additionally fits imputation and standardization on Train. The model-neutral prediction helper supports either final family.

```python
# PROVIDED METRICS — DO NOT CHANGE
MAIN_METRIC = "Macro-F1"
def evaluate_scores(y_true, probabilities, threshold=0.5):
    probabilities = np.asarray(probabilities)
    assert probabilities.shape == (len(y_true),)
    assert np.isfinite(probabilities).all()
    predictions = (probabilities >= threshold).astype(int)
    return {
        "Accuracy": accuracy_score(y_true, predictions),
        "Macro-F1": f1_score(y_true, predictions, labels=[0, 1], average="macro", zero_division=0),
        "Balanced Accuracy": balanced_accuracy_score(y_true, predictions),
        "Fail Precision": precision_score(y_true, predictions, zero_division=0),
        "Fail Recall": recall_score(y_true, predictions, zero_division=0),
        "Fail F1": f1_score(y_true, predictions, zero_division=0),
        "AP": average_precision_score(y_true, probabilities),
    }

def fit_preprocessor(X_fit, mode):
    if mode not in {"native", "mean", "median"}:
        raise ValueError("mode must be native, mean, or median")
    columns = X_fit.columns[X_fit.nunique(dropna=True) > 1].tolist()
    assert columns, "No usable training features."
    imputer = None
    if mode != "native":
        imputer = SimpleImputer(strategy=mode)
        imputer.fit(X_fit[columns])
    return {"mode": mode, "columns": columns, "imputer": imputer}

def transform_features(X_input, preprocessing):
    values = X_input[preprocessing["columns"]]
    if preprocessing["imputer"] is not None:
        values = preprocessing["imputer"].transform(values)
    return np.asarray(values, dtype=np.float32)

BASE_PARAMS = dict(
    objective="binary:logistic", booster="gbtree", tree_method="hist",
    eval_metric="logloss", n_estimators=150, learning_rate=0.05,
    max_depth=3, min_child_weight=1, subsample=0.8, colsample_bytree=0.8,
    reg_alpha=0.0, reg_lambda=1.0, scale_pos_weight=1.0,
    random_state=TRAIN_SEED, n_jobs=2,
)

def train_candidate(mode, overrides):
    preprocessing = fit_preprocessor(X_train, mode)
    params = {**BASE_PARAMS, **overrides}
    model = XGBClassifier(**params)
    model.fit(transform_features(X_train, preprocessing), y_train)
    val_prob = model.predict_proba(transform_features(X_val, preprocessing))[:, 1]
    return {"model": model, "preprocessing": preprocessing, "params": params,
            "mode": mode, "val_prob": val_prob, "family": "XGBoost"}

# EDITABLE MLP TRAINING — loss, architecture, and training strategy may be changed.
from torch.utils.data import WeightedRandomSampler
from sklearn.preprocessing import RobustScaler, QuantileTransformer
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.impute import KNNImputer
MLP_BASE_PARAMS = dict(hidden_sizes=(128, 64), dropout=0.3, epochs=60,
    batch_size=64, learning_rate=1e-3, weight_decay=1e-4,
    pos_weight=1.0, seed=TRAIN_SEED, loss="bce", focal_gamma=2.0,
    max_missing=1.0, select_by=MAIN_METRIC, scaler="standard", clip_value=None,
    corr_threshold=None, select_k=None, missing_indicator=False, input_noise=0.0,
    knn_neighbors=None, ema_decay=None, lr_schedule=None, balanced_sampler=False,
    input_dropout=0.0, batch_norm=False, activation="relu")

def make_mlp(input_dim, params):
    layers = [nn.Dropout(params["input_dropout"])] if params["input_dropout"] > 0 else []
    for width in params["hidden_sizes"]:
        layers += [nn.Linear(input_dim, width)]
        if params["batch_norm"]:
            layers += [nn.BatchNorm1d(width)]
        layers += [nn.GELU() if params["activation"] == "gelu" else nn.ReLU(),
                   nn.Dropout(params["dropout"])]
        input_dim = width
    return nn.Sequential(*layers, nn.Linear(input_dim, 1))
    
def mlp_loss(logits, targets, params):
    # Replace this function freely with another suitable binary classification loss.
    if params["loss"] == "bce":
        return nn.functional.binary_cross_entropy_with_logits(logits, targets,
            pos_weight=logits.new_tensor(params["pos_weight"]))
    if params["loss"] == "focal":
        bce = nn.functional.binary_cross_entropy_with_logits(logits, targets, reduction="none")
        prob = torch.sigmoid(logits)
        pt = torch.where(targets == 1, prob, 1 - prob)
        weights = torch.where(targets == 1, params["pos_weight"], 1.0)
        return (weights * (1 - pt).pow(params["focal_gamma"]) * bce).mean()
    raise ValueError("Implement your custom loss here or use bce/focal.")

def mlp_features_base(X_input, preprocessing):
    if preprocessing.get("knn_imputer") is not None:
        return preprocessing["knn_imputer"].transform(
            X_input[preprocessing["columns"]].to_numpy(np.float32)).astype(np.float32)
    return transform_features(X_input, preprocessing)

def fit_mlp_features(train_raw, preprocessing, params):
    if params["knn_neighbors"] is not None:
        preprocessing["knn_imputer"] = KNNImputer(n_neighbors=params["knn_neighbors"]).fit(
            train_raw[preprocessing["columns"]].to_numpy(np.float32))
    values = mlp_features_base(train_raw, preprocessing)
    if params["missing_indicator"]:
        preprocessing["indicator_columns"] = [
            c for c in preprocessing["columns"] if train_raw[c].isna().any()]
        values = np.hstack([values,
            train_raw[preprocessing["indicator_columns"]].isna().to_numpy(np.float32)])
    keep = np.arange(values.shape[1])
    if params["corr_threshold"] is not None:
        corr = np.abs(np.nan_to_num(np.corrcoef(values, rowvar=False)))
        keep = np.where(~(np.triu(corr, k=1) > params["corr_threshold"]).any(axis=0))[0]
    if params["select_k"] is not None and params["select_k"] < len(keep):
        selector = SelectKBest(f_classif, k=params["select_k"]).fit(values[:, keep], y_train)
        keep = keep[selector.get_support()]
    preprocessing["keep_index"] = keep
    scaler = {"standard": StandardScaler(), "robust": RobustScaler(),
              "quantile": QuantileTransformer(output_distribution="normal",
                  n_quantiles=min(1000, len(values)), random_state=0)}[params["scaler"]]
    preprocessing["scaler"] = scaler.fit(values[:, keep])
    preprocessing["clip_value"] = params["clip_value"]
    return mlp_features(train_raw, preprocessing)

def mlp_features(X_input, preprocessing):
    values = mlp_features_base(X_input, preprocessing)
    if preprocessing.get("indicator_columns"):
        values = np.hstack([values,
            X_input[preprocessing["indicator_columns"]].isna().to_numpy(np.float32)])
    if "keep_index" in preprocessing:
        values = values[:, preprocessing["keep_index"]]
    values = preprocessing["scaler"].transform(values)
    if preprocessing.get("clip_value") is not None:
        values = np.clip(values, -preprocessing["clip_value"], preprocessing["clip_value"])
    return values

def train_mlp(mode="median", overrides=None):
    params = {**MLP_BASE_PARAMS, **(overrides or {})}
    torch.manual_seed(params["seed"])
    # CPU baseline; device selection is editable. A small thread count suits this dataset.
    torch.set_num_threads(2)
    kept_columns = X_train.columns[X_train.isna().mean() <= params["max_missing"]]
    preprocessing = fit_preprocessor(X_train[kept_columns], mode)
    assert mode in {"mean", "median"}, "MLP requires finite inputs; impute first."
    tx = torch.tensor(fit_mlp_features(X_train, preprocessing, params), dtype=torch.float32)
    vx = torch.tensor(mlp_features(X_val, preprocessing), dtype=torch.float32)
    ty = torch.tensor(y_train.to_numpy(), dtype=torch.float32)
    assert torch.isfinite(tx).all() and torch.isfinite(vx).all()
    if params["balanced_sampler"]:
        class_counts = torch.bincount(ty.long()).float()
        sampler = WeightedRandomSampler((1.0 / class_counts)[ty.long()], num_samples=len(ty),
            replacement=True, generator=torch.Generator().manual_seed(params["seed"]))
        loader = DataLoader(TensorDataset(tx, ty), batch_size=params["batch_size"], sampler=sampler)
    else:
        loader = DataLoader(TensorDataset(tx, ty), batch_size=params["batch_size"], shuffle=True,
            generator=torch.Generator().manual_seed(params["seed"]))
    model = make_mlp(tx.shape[1], params)
    optimizer = torch.optim.AdamW(model.parameters(), lr=params["learning_rate"],
                                 weight_decay=params["weight_decay"])
    scheduler = (torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=params["epochs"])
                 if params["lr_schedule"] == "cosine" else None)
    ema_state = ({k: v.detach().clone() for k, v in model.state_dict().items()}
                 if params["ema_decay"] is not None else None)
    eval_model = copy.deepcopy(model) if ema_state is not None else model
    history, best_score, best_state, best_epoch = [], -1.0, None, None
    for epoch in range(params["epochs"]):
        model.train()
        total_loss = 0.0
        for xb, yb in loader:
            if params["input_noise"] > 0:
                xb = xb + params["input_noise"] * torch.randn_like(xb)
            optimizer.zero_grad()
            loss = mlp_loss(model(xb).squeeze(1), yb, params)
            loss.backward()
            optimizer.step()
            if ema_state is not None:
                with torch.no_grad():
                    for k, v in model.state_dict().items():
                        if v.dtype.is_floating_point:
                            ema_state[k].mul_(params["ema_decay"]).add_(v, alpha=1 - params["ema_decay"])
                        else:
                            ema_state[k].copy_(v)
            total_loss += loss.item() * len(yb)
        if scheduler is not None:
            scheduler.step()
        if ema_state is not None:
            eval_model.load_state_dict(ema_state)
        eval_model.eval()
        with torch.no_grad():
            val_prob = torch.sigmoid(eval_model(vx).squeeze(1)).numpy()
        # Baseline checkpoint selection: Validation Macro-F1 at threshold 0.5.
        # You may change checkpoint selection, but never use Test for selection.
        score_table = evaluate_scores(y_val, val_prob)
        score = score_table[params["select_by"]]
        history.append({"Epoch": epoch + 1, "Train loss": total_loss / len(ty),
                        "Validation Macro-F1 @0.5": score_table[MAIN_METRIC],
                        "Validation AP": score_table["AP"]})
        if score > best_score:
            best_score, best_epoch = score, epoch + 1
            best_state = copy.deepcopy(eval_model.state_dict())
    assert best_state is not None
    model.load_state_dict(best_state)
    model.eval()
    with torch.no_grad():
        val_prob = torch.sigmoid(model(vx).squeeze(1)).numpy()
    return {"model": model, "preprocessing": preprocessing, "params": params,
            "mode": mode, "family": "MLP", "val_prob": val_prob,
            "history": history, "best_epoch": best_epoch}

def predict_candidate(candidate, X_input):
    values = transform_features(X_input, candidate["preprocessing"])
    if candidate["family"] == "XGBoost":
        return candidate["model"].predict_proba(values)[:, 1]
    values = mlp_features(X_input, candidate["preprocessing"])
    candidate["model"].eval()
    with torch.no_grad():
        return torch.sigmoid(candidate["model"](
            torch.tensor(values, dtype=torch.float32)).squeeze(1)).cpu().numpy()
```

### 2. 이론적 분석

#### XGBoost

XGBoost는 decision tree를 순차적으로 더해 가며 예측값을 만드는 gradient boosting model이다.

$$
F_M(x)=\sum_{m=1}^{M} f_m(x),\qquad p(y=1\mid x)=\sigma(F_M(x))
$$

$m$번째 tree는 이전까지의 예측이 만든 loss를 줄이도록 학습된다. XGBoost는 loss를 2차 Taylor 전개로 근사하고 tree의 복잡도에 대한 regularization을 함께 최소화한다.

$$
\mathcal{L}^{(m)} \approx \sum_{i}\left[g_i f_m(x_i) + \frac{1}{2}h_i f_m(x_i)^2\right] + \gamma T + \frac{1}{2}\lambda\sum_{j=1}^{T} w_j^2
$$

여기서 $g_i$와 $h_i$는 loss의 1차 및 2차 미분이고 $T$는 leaf 수, $w_j$는 leaf 값이다. 각 leaf의 최적값과 분기의 gain은 다음과 같이 계산된다.

$$
w_j^* = -\frac{G_j}{H_j+\lambda},\qquad
\mathrm{Gain}=\frac{1}{2}\left[\frac{G_L^2}{H_L+\lambda}+\frac{G_R^2}{H_R+\lambda}-\frac{(G_L+G_R)^2}{H_L+H_R+\lambda}\right]-\gamma
$$

XGBoost는 결측값이 있는 sample을 왼쪽과 오른쪽 중 gain이 더 큰 방향으로 보내는 default direction을 학습한다. 따라서 결측치를 따로 채우지 않아도 결측 여부 자체를 분기 정보로 사용할 수 있다.

`scale_pos_weight`는 불량 sample의 gradient와 hessian에 가중치를 곱하여 불량 sample이 loss에서 차지하는 비중을 키운다. 이는 model이 출력하는 확률의 크기를 불량 쪽으로 이동시키며 같은 threshold에서 불량으로 예측되는 sample 수를 증가시킨다.

#### MLP

MLP는 표준화된 sensor vector를 여러 개의 fully connected layer와 비선형 activation에 통과시켜 하나의 불량 logit $z$를 출력한다.

$$
h^{(l)} = \mathrm{Dropout}\left(\mathrm{ReLU}\left(W^{(l)}h^{(l-1)}+b^{(l)}\right)\right),\qquad
z = W^{(L)}h^{(L-1)}+b^{(L)},\qquad \hat{p}=\sigma(z)
$$

불균형을 보정하기 위해 불량 sample의 loss에 `pos_weight` $w_+$를 곱한 weighted BCE를 사용하였다.

$$
\mathcal{L}_{\mathrm{BCE}} = -\frac{1}{N}\sum_{i=1}^{N}\left[w_+\, y_i\log\hat{p}_i + (1-y_i)\log(1-\hat{p}_i)\right]
$$

$w_+ = N_{pass}/N_{fail}$로 설정하면 불량 73개와 정상 1,023개가 loss에 거의 같은 비중으로 기여한다.

Dropout은 학습 중 hidden unit의 일부를 무작위로 제거하여 특정 unit의 조합에 의존하지 않도록 한다. AdamW의 weight decay는 weight의 크기를 줄여 과적합을 억제한다.

#### 평가 지표와 Decision Threshold

불량 비율이 약 6.6%이므로 모든 sample을 정상으로 예측해도 Accuracy는 약 0.93이다. 따라서 이번 실습에서는 두 class의 F1을 같은 비중으로 평균한 Macro-F1을 주 지표로 사용하였다.

$$
\mathrm{Macro\text{-}F1}=\frac{F1_{pass}+F1_{fail}}{2}
$$

All-pass 예측은 $F1_{fail}=0$이므로 Accuracy가 높아도 Macro-F1은 약 0.48에 머문다.

Average Precision(AP)은 threshold를 정하지 않고 불량 sample에 정상 sample보다 높은 score를 주는 능력을 평가한다. 무작위 예측의 AP는 불량 비율과 같은 약 0.067이다.

Macro-F1은 threshold에 따라 달라진다. Threshold를 낮추면 불량 Recall은 증가하지만 정상 sample을 불량으로 잘못 예측하는 False Positive도 증가한다. 이번 실습에서는 Validation에서 0.01부터 0.99까지 threshold를 탐색하였으며 Macro-F1이 같을 경우 0.5에 가까운 값을 선택하였다.

#### 상관된 Sensor 제거와 Batch Size

SECOM에는 같은 공정을 측정하는 sensor가 많아 서로 거의 같은 값을 가지는 sensor 쌍이 많다. Train sample이 1,096개이고 불량이 73개뿐인 상황에서 중복된 sensor가 많으면 MLP가 같은 정보에 해당하는 많은 weight를 동시에 학습하므로 적은 수의 불량 sample에 과적합되기 쉽다. Train에서 상관계수 0.95를 넘는 sensor를 제거하면 정보는 거의 유지하면서 입력 차원을 줄일 수 있다.

Mini-batch gradient의 분산은 batch size $B$에 반비례한다.

$$
\mathrm{Var}\left[\hat{\nabla}_B\mathcal{L}\right] \approx \frac{\sigma^2}{B}
$$

Batch size를 줄이면 epoch당 update 횟수가 증가하고 gradient에 더 큰 noise가 포함된다. 이 noise는 parameter가 Train에 지나치게 맞춰진 좁은 최솟값으로 수렴하는 것을 방해하여 regularization과 비슷한 효과를 낼 수 있다.

5차 실험에서는 효과가 있던 전처리를 조합하였지만 추가적인 향상은 없었다. 특히 RobustScaler와 clipping은 단독으로 사용할 때와 달리 상관 sensor 제거와 함께 사용하면 크게 낮아졌다. Corr0.9는 평균이 가장 높았지만 threshold를 평균 0.76까지 높여야 했고 threshold 0.5에서는 corr0.95보다 낮았으며 표준편차도 컸다. 결측치 대체 방법은 중앙값 대체가 평균 대체와 KNN 대체보다 좋았다. SECOM sensor에는 극단적인 값이 많아 평균은 이상치의 영향을 받고, KNN은 값의 범위가 큰 sensor가 이웃 선택을 좌우하기 때문으로 판단하였다.

6차 실험에서는 전처리를 corr0.95로 고정하고 학습 방법을 비교하였다. Batch size를 32로 줄였을 때 Macro-F1 평균이 0.738로 가장 높았고 세 seed의 결과가 모두 0.729 이상으로 표준편차도 0.009로 작았다. 반면 가중치 EMA는 seed에 따른 변동은 크게 줄였지만(표준편차 0.008) 평균은 0.696으로 낮아졌다. EMA가 epoch마다 갑자기 높아지는 Validation 점수까지 평균으로 없애기 때문이며 이번 실험의 점수 변동은 주로 Validation 불량 10개에서 발생하는 noise였기 때문에 평균 성능은 향상되지 않았다. Balanced sampler는 불량 Precision이 높아지고 Recall이 낮아지는 경향을 보였다.

#### 최종 Model 선택 방법

전체 실험 중 Validation Macro-F1이 가장 높은 단일 결과는 corr0.9의 seed 0(0.755, threshold 0.85)이었다. 하지만 100개가 넘는 실험 중에서 가장 높은 단일 결과를 선택하면 Validation에 우연히 잘 맞은 model을 고를 가능성이 크다.

따라서 다음 두 단계로 최종 model을 선택하였다.

1. Seed 평균 Validation Macro-F1이 가장 높고 표준편차가 작은 설정(corr0.95 + batch size 32)을 선택하였다.
2. 해당 설정의 seed 중 Validation Macro-F1이 가장 높은 model(seed 0)을 최종 model로 선택하였다.

최종 model의 Validation Macro-F1은 0.7479로 단일 최고 결과(0.755)와 0.007 차이였으며 threshold를 조정하지 않은 0.5에서 이 값을 얻었다. 두 단계 모두 Validation만 사용하였으며 선택한 설정, 가중치 그리고 threshold를 고정한 이후 FINAL TEST EVALUATION cell을 한 번만 실행하였다.

#### Confusion Matrix

Validation에서는 불량 10개 중 5개를 검출하고 정상 147개 중 4개만 불량으로 잘못 판정하였다. Test에서는 정상 293개 중 282개(96.2%)를 올바르게 판정하였지만 불량은 21개 중 3개만 검출하였다. 불량 경보 14개 중 3개만 실제 불량이었으므로 Fail Precision은 0.214이다.

Validation으로 선택한 최고 XGBoost는 Test에서 불량 경보를 5개만 내고 그중 1개만 실제 불량이었다. XGBoost는 정상 판정에 치우쳐 불량을 거의 검출하지 못한 반면 최종 MLP는 False Positive를 더 허용하는 대신 더 많은 불량을 검출하였다.

실제 공정에서는 불량을 놓치는 False Negative가 다음 공정으로 불량품을 넘기는 비용을 만들고, False Positive는 정상 제품을 재검사하는 비용을 만든다. 일반적으로 불량을 놓치는 비용이 더 크므로 최종 model의 Fail Recall 0.143은 실제 사용에는 아직 부족한 수준이다.

#### Threshold에 따른 변화

Validation threshold 그래프에서 Macro-F1은 threshold가 증가할수록 높아져 0.4\~0.5 부근에서 최대값 0.748에 도달하였다. Threshold가 0.5를 넘으면 Fail Recall이 0.5에서 0.4, 0.3으로 감소하면서 Macro-F1도 낮아졌다.

특이한 점은 threshold를 0.01까지 낮추어도 Fail Recall이 0.5에서 더 증가하지 않았다는 것이다. 즉 Validation 불량 10개 중 5개는 score가 0.01보다 작았다. Train loss가 거의 0까지 감소한 것처럼 model이 Train에 강하게 맞춰지면서 확률 출력이 0 또는 1에 가깝게 치우쳤고, model이 정상이라고 확신한 불량 sample은 threshold를 조정해도 검출할 수 없었다.

#### 학습 곡선

최종 model의 Train loss는 약 1.3에서 시작하여 20 epoch 무렵 0.05 이하로 감소하였다. Validation Macro-F1은 epoch 18에서 최대값 0.748에 도달한 뒤 0.57\~0.70 사이에서 변동하였고 Validation AP도 epoch 15 부근에서 최대값을 기록한 뒤 감소하였다. 따라서 epoch 18 이후에는 model이 Train에 과적합되기 시작한 것으로 볼 수 있으며 Validation Macro-F1로 checkpoint를 선택한 것이 과적합 이전의 model을 사용하는 역할을 하였다.

#### 높은 Threshold의 위험

팀의 다른 실험에서는 Validation에서 threshold가 0.98로 선택된 MLP를 최종 model로 사용하였다. 해당 model은 Test에서 불량 21개 중 1개만 검출하여 Test Macro-F1 0.5185를 기록하였다. Validation 불량 10개에 맞춰 threshold를 극단적으로 조정하면 Test로 일반화되지 않을 수 있음을 보여주며, threshold 0.5 근처에서 안정적인 설정을 선택한 방법이 Test에서 더 좋은 결과로 이어졌다.


### 6. 한계 및 개선 방향

첫 번째 한계는 Validation set의 크기이다.

Validation 불량이 10개뿐이므로 불량 하나에 따라 Macro-F1이 크게 변하고, 많은 설정 중에서 가장 높은 점수를 선택하면 Validation 점수가 실제 성능보다 높게 평가된다. 이번 실험에서도 Validation 0.748이 Test에서는 0.561로 감소하였다. 이를 완화하기 위해 seed 평균으로 설정을 비교하였지만 근본적인 해결책은 아니다. 향후에는 Train 내부에서 stratified K-fold cross-validation을 수행하여 더 많은 불량 sample로 설정을 비교하고 Validation은 최종 확인에만 사용하는 방법을 사용할 수 있다.

두 번째는 데이터 자체의 한계이다.

Train 불량은 73개뿐이고 모든 model의 Test AP가 0.15\~0.19 수준에 머물렀다. 이는 현재 sensor feature만으로는 불량과 정상을 잘 구분하기 어렵다는 것을 의미한다. 실제 공정에서는 더 많은 불량 data를 확보하거나 공정 단계, 장비 정보, 시간 정보와 같은 추가 feature를 사용하는 것이 model 변경보다 더 큰 효과를 낼 것으로 생각한다. 또한 이번 실습은 random split을 사용하였으므로 미래 시점의 생산 data에 대한 성능은 평가하지 못하였다.

세 번째는 model의 과신 경향이다.

Threshold 분석에서 Validation 불량의 절반이 0.01보다 낮은 score를 받았다. 향후에는 label smoothing이나 temperature scaling과 같은 calibration 방법을 적용하여 확률 출력이 덜 극단적이 되도록 하면 threshold 조정의 효과가 커질 수 있다.

네 번째는 실험 설계의 한계이다.

각 설정을 seed 3개로만 비교하였으므로 평균 차이가 0.01 이하인 설정 사이의 우열은 확실하지 않다. 또한 4\~6차 실험은 이전 단계의 최고 설정을 기준으로 진행하였으므로 다른 기준 설정에서는 다른 결과가 나올 수 있다.

마지막으로 XGBoost는 tree 수, 깊이, sampling 비율 정도만 조정하였다. Learning rate와 tree 수를 함께 더 넓게 탐색하거나 MLP에서 효과가 있었던 상관 sensor 제거를 XGBoost에도 적용했다면 두 model의 차이가 달라졌을 가능성이 있다.


### 7. 팀 작업 과정

팀에서는 먼저 제공된 notebook의 고정 분할, label mapping 그리고 평가 함수를 그대로 유지한 상태에서 XGBoost와 MLP baseline을 실행하고 결과를 함께 확인하였다. XGBoost가 threshold 0.5에서 불량을 하나도 예측하지 않지만 AP는 All-pass보다 높다는 점, MLP가 빠르게 과적합된다는 점을 공유하고 이를 바탕으로 실험 방향을 정하였다.

1차 실험에서는 불균형 보정, 결측치 처리 방식 그리고 loss function을 두 model에 나누어 적용하였다. MLP가 더 높은 결과를 보였지만 XGBoost가 충분히 조정되지 않았다는 의견이 있었고, 일반적으로 표 형태의 data에서는 tree 기반 model이 유리하다는 점을 고려하여 XGBoost도 추가로 조정한 뒤 비교하기로 하였다.

2차 실험 이후에는 Validation 불량이 10개뿐이라 실험 사이의 차이가 sample 한두 개 차이라는 점을 확인하였다. 따라서 단일 결과로 model을 결정하지 않고 두 model의 최고 설정을 여러 seed로 학습하여 평균을 비교하기로 하였으며 그 결과를 근거로 MLP를 최종 model family로 선택하였다.

MLP를 선택한 이후에는 전처리, 결측치 처리 그리고 학습 방법을 단계적으로 실험하였다. 각 단계에서 효과가 있었던 설정을 다음 단계의 기준으로 사용하였으며 모든 비교는 seed 3개의 평균으로 수행하였다. 실험 중에는 Validation 단일 최고 결과를 그대로 선택할지, seed 평균으로 설정을 먼저 정할지 논의하였고 Validation 과적합을 줄이기 위해 두 단계 선택 방법을 사용하기로 하였다.

팀의 다른 실험에서는 Validation에서 threshold 0.98이 선택된 MLP를 최종 model로 사용하였다. 두 결과를 비교하면서 threshold를 Validation에 지나치게 맞추는 것의 위험을 확인하였다.

모든 model 비교, threshold 선택 그리고 최종 model 결정은 Validation만 이용하였다. 최종 model을 고정한 이후에만 FINAL TEST EVALUATION cell을 한 번 실행하였고 Test 결과를 확인한 이후에는 어떠한 설정도 변경하지 않았다.


### 8. 결론

이번 실습에서는 결측치와 클래스 불균형이 있는 SECOM 반도체 공정 data에서 XGBoost와 MLP를 학습하고 Validation Macro-F1을 기준으로 비교하였다.

두 model 모두 기본 설정에서는 불량을 거의 검출하지 못하였다. 불량 가중치와 threshold 조정을 통해 불량 검출이 가능해졌으며 두 model의 하이퍼파라미터를 모두 조정한 뒤 seed 5개로 비교한 결과 MLP가 평균 Validation Macro-F1 0.700으로 XGBoost(0.661)보다 높아 MLP를 최종 model family로 선택하였다.

MLP에서는 상관계수 0.95를 넘는 중복 sensor를 제거하는 전처리와 batch size 32를 사용했을 때 seed 평균 Validation Macro-F1이 0.738 ± 0.009로 가장 높고 안정적이었다. 반면 결측 지시 변수, QuantileTransformer, 강한 feature 선택, BatchNorm, cosine schedule 그리고 가중치 EMA는 성능을 높이지 못하였다.

최종 model은 Validation Macro-F1 0.7479를 기록하였으며 held-out Test에서 한 번 평가한 결과 Final Test Macro-F1 0.5613을 얻었다. 이는 All-pass(0.4827), MLP baseline(0.5140) 그리고 Validation으로 선택한 최고 XGBoost(0.5185)보다 높은 결과이다.

하지만 Validation과 Test 사이에 약 0.19의 차이가 있었고 Test에서 불량 21개 중 3개만 검출하였다. 이를 통해 불량 sample이 매우 적은 data에서는 Validation 점수가 실제 성능을 과대평가하기 쉽다는 점을 확인하였다. 따라서 여러 seed의 평균으로 설정을 비교하고 threshold를 극단적으로 조정하지 않는 것처럼 Validation 과적합을 줄이는 model 선택 방법이 중요하다는 점을 확인하였다.
